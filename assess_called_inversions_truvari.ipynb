# 05 Assess the Called Inversions
Visualizing the data was useful, but we need a more rigorous solution to evaluate the results. Specifically, we are interested in the false and true positive/negative rates for each of the treatments. We can do that using [truvari](https://github.com/ACEnglish/truvari?tab=readme-ov-file).

```python
import os
```

```python
def convert_to_vcfgz(infile:str):
    newname = ".".join(infile.split(".")[:-1]) + ".vcf.gz"
    os.system(f"~/.pixi/bin/pixi run -e harpy bcftools view -Oz -o {newname} --write-index=tbi {infile}")
    return newname

def truvari(truth_vcf: str, comparison_bcf: str, outdir:str):
    com_vcf = convert_to_vcfgz(comparison_bcf)
    #os.system(f"~/.pixi/bin/pixi run -e harpy bcftools view -Oz -o {com_vcf} {ref_vcf}")
    os.system(
        f"~/.pixi/bin/pixi run -e harpy truvari bench -b {truth_vcf} -c {com_vcf} -o {outdir}"
    )
```

This requires a non-gzipped version of the genome

```python
if not os.path.exists("dmel_genome/dmel_2_3.fa.gz"):
    os.system("gzip -dk dmel_genome/dmel_2_3.fa.gz")
```

```python
size = "large"
_depth = "depth_20"
_sample = "sample_" + "3".zfill(2)
os.makedirs(f"assess_results/{size}/{_depth}", exist_ok=True)
truvari(
    truth_vcf = f"simulated_variants/inversions_genomes/{size}/samples/{_sample}/{_sample}.{size}.vcf.gz",
    comparison_bcf = f"simulated_data/called_sv/leviathan/{size}/{_depth}/by_sample90/vcf/{_sample}.bcf",
    outdir = f"assess_results/{size}/{_depth}/by_sample90"
)
```
